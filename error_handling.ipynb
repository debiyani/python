```python
try:
    a=int(input('Enter first number: '))
    b=int(input('Enter second number: '))
    c=a/b
except ValueError: # like catch used to show error
    print("Error: Only whole numbers can be sent to input field.")
except ZeroDivisionError:
    print("Error: Number cannot be divisible by 0.")
else: # used to print output if no error
    print(c)
finally: #runs regardless of error or not
    print('Program ended.')
```
