```python
# append
fruits.append('Coconut')
fruits
```

```python
numbers=[1,1,1,2,2,4,6,3,5,5]
duplicate_data=[]
new=[]
duplicate=[]
# only using list, loop and append separate only duplicate data in duplicate[]
for index, num  in enumerate(numbers):
    if num in new:
        duplicate_data.append(num)
    else:
        new.append(num)
for index, num in enumerate(duplicate_data):
    if num in duplicate:
        continue
    else:
        duplicate.append(num)
print(duplicate_data)
print(f"Numbers: {numbers}")
print(f"Unique items: {new}")
print(f"Duplicate data: {duplicate}")

```

```python
numbers=[1,2,3,4,5,6,7,1,1,1,1,1,2,2]
data=[]
duplicate_data=[]

for num in numbers:
    if num in data:
        if num not in duplicate_data:
            duplicate_data.append(num)
    else:
        data.append(num)
```
