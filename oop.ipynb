### Inheritance

##### Single Inheritance

```python
'''  
1. Parent class and child class
2. child class inherits/extends parent class
3. If child class inherits parent class then child class must inherit parent class constructor
  -> reason:
'''
```

```python
# parent class
class Customers:
    def __init__(self, customer_id, customer_name, customer_email, customer_phone):
        self.customer_id=customer_id
        self.customer_name= customer_name
        self.customer_email= customer_email
        self.customer_phone= customer_phone
    def customer_details(self):
        print(f'Customer ID: {self.customer_id}')
        print(f'Customer Name: {self.customer_id}')
        print(f'Customer Email: {self.customer_id}')
        print(f'Customer Phone: {self.customer_id}')
```

```python
# child class
class Orders(Customers):
    def __init__(self, order_id, order_date, order_status, order_location, customer_id, customer_name, customer_email, customer_phone):
        # constructor inheritance
        super().__init__(customer_id, customer_name, customer_email, customer_phone)
        self.order_id = order_id
        self.order_date = order_date
        self.order_status = order_status
        self.order_location = order_location
        
    def order_details(self):
        print(f'Order ID: {self.order_id}')
        print(f'Order Date: {self.order_date}')
        print(f'Order Status: {self.order_status}')
        print(f'Order Location: {self.order_location}')
```

```python
o= Orders(1,'2026-08-12','Processing','Bhaktapur','101','Ram','ram@gmail.com','98933334532')
```

```python
o.order_details()
```

```python
o.customer_details()
```

```python
# multilevel inheritance
class Order_Items(Orders):
    def __init__(self,item_id, quantity, list_price, discount, order_id, order_date, order_status, order_location, customer_id, customer_name, customer_email, customer_phone):
        super().__init__(order_id, order_date, order_status, order_location, customer_id, customer_name, customer_email, customer_phone )
        
        self.item_id=item_id
        self.quantity=quantity
        self.list_price=list_price
        self.discount=discount
        
    def order_item_details(self):
        print(f'Item ID: {self.item_id}')
        print(f'Item quantity: {self.quantity}')
        print(f'Item price: {self.list_price}')
        print(f'Item Discount: {self.discount}')
        
```

```python
or_it=Order_Items('101', 2, 500, 10,1,'2026-08-12','Processing','Bhaktapur','101','Ram','ram@gmail.com','98933334532')
```

```python
or_it.customer_details()
```

```python
or_it.order_details()
```

```python
or_it.order_item_details()
```

### Multiple Inheritance

```python
class Category:
    def __init__(self, category_id, category_name):
        self.category_id=category_id
        self.category_name=category_name
        
    def category_details(self):
        print(f'Category_id: {self.category_id}')
        print(f'Category_name: {self.category_name}')


```

```python
class Brand:
    def __init__(self, brand_id, brand_name):
        self.brand_id=brand_id
        self.brand_name=brand_name
        
    def brand_details(self):
        print(f'Category_id: {self.category_id}')
        print(f'Category_name: {self.category_name}')
```

```python
class Product(Category, Brand):
    def __init__(self, product_id, product_name, model_year, list_price,category_id, category_name, brand_id, brand_name):
        Category.__init__(self, category_id, category_name)
        Brand.__init__(self, brand_id, brand_name)
        self.product_id=product_id
        self.product_name=product_name
        self.model_year=model_year
        self.list_price= list_price
        
    def product_details(self):
        print(f'Product ID: {self.product_id}')
        print(f'Product Name: {self.product_name}')
        print(f'Product Year: {self.model_year}')
        print(f'Product Price: {self.list_price}')
```

```python
p=Product('p101','Bottle',2018,400,1,'Utensils','b1','always')
```

```python
p.product_details()
```

```python
p.category_details()
```

```python
p.brand_details()
```

```python
# child class
class Orders(Customers):
    def __init__(self, order_id, order_date, order_status, order_location, customer_id, customer_name, customer_email, customer_phone):
        # constructor inheritance
        super().__init__(customer_id, customer_name, customer_email, customer_phone)
        self.__order_id = order_id
        self.order_date = order_date
        self._order_status = order_status
        self.order_location = order_location
        
    def order_details(self):
        print(f'Order ID: {self.__order_id}')
        print(f'Order Date: {self.order_date}')
        print(f'Order Status: {self.order_status}')
        print(f'Order Location: {self.order_location}')
```

### Polymorphism

```python
## messaging
class Messenger:
    def message(self, send_from, to, message):
        print(f'Messenger: {to}')
        print(f'{send_from} sent you a message.')
        print(f'Message :{message}')
        
class Whatsapp:
    def message(self, send_from, to, message):
            print(f'WhatsApp: {to}')
            print(f'{send_from} sent you a message.')
            print(f'Message :{message}')
            
class Viber:
    def message(self, send_from, to, message):
            print(f'Viber: {to}')
            print(f'{send_from} sent you a message.')
            print(f'Message :{message}')
```

```python
def send_message(obj, send_from, to, message):
    obj.message(send_from, to, message)
```

```python
send_message(Whatsapp(), 'Ram', 'Shyam','Hahah!!!')
```

```python
send_from = input('Enter your name: ')
to = input('Enter receiver name: ')
message = input('Enter your message: ')

print('On which platform do oyu want to send message?: ')
print('1. Messenger')
print('2. WhatsApp')
print('3. Viber')

choice = int(input('Enter your choice: '))

if choice==1:
    obj=Messenger()
elif choice==2:
    obj=Whatsapp()
elif choice==3:
    obj=Viber()
else:
    print('Invalid choice.')

send_message(obj, send_from, to, message)
```

### Abstraction

```python
''' 
Implementation hiding
1. A class needs to be an abstract class.
    -> have to inherit Abstract Base Class (ABC) by importing ABC from abc.
2. Abstract class must require atleast 1 abstract function.
'''
```

```python
# abstract base class
from abc import ABC, abstractmethod
```

```python
class Bank(ABC):
    @abstractmethod
    def userDetails(self):
        pass
```

```python
class User(Bank):
    def personalDetails(self):
        print('Name:')
        print('Address:')
        
    def userDetails(self):
        print('Balance: Rs. 50000')
        print('A/C no.: 123321123454321')
```

```python
u=User()
```

```python
u.userDetails()
```
