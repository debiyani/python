```python
prices = [100, 250, 75, 400, 120, 310]
cheap_items=[]
expensive_items=[]
for price in prices:
    if price>200:
        expensive_items.append(price)
    else:
        cheap_items.append(price)
print(f"Expensive items: {expensive_items}")
print(f"Cheap items: {cheap_items}")
print(f"Totak count of expensive items: {len(expensive_items)}")
print(f"Total count of cheap items: {len(cheap_items)}")
```

### Tuple

```python
# immutable(i.e. we can not change, update or delete)
# two ways to define a tuple
a_tuple=(1,2,3,4,5)
print(a_tuple)
# usually denoted by ()
```

```python
# can also define tuple in the following way and python wont show any errror
b_tuple =1,2,3,4
print(type(b_tuple))
```

```python
for index,data in enumerate(b_tuple):
    print(f"{index}:{data}")
```

```python
# can access value in a tuple but cant remove update add
# can convert to list to update a tuple
# no function to update or remove data
data=list(b_tuple)
data[3]=10
b_tuple=tuple(data)
b_tuple
```
