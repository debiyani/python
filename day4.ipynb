### Match Case

```python
# similar to switch case
# default case is reoresented by _(eg, Case _:)
'''
Syntax:
match value:
    Case 1:
        //code
    Case 2:
        //code
    Case n:
        //code
    Case _:
        //code
'''
```

#### Restaurant Order System using match case

```python
# Note if else faster than match case
# match case compares exact value by converting it to binary
```

```python
print("Welcome to Restaurant Order System")
print("1. Breakfast")
print("2. Lunch")
print("3. Snacks")
print("4. Dinner")
print("5. Dessert")
print("6. Drinks")

choice = int(input("Enter your order choice (1-6):"))

match choice:
    case 1:
        print("You have ordered breakfast.")
        print("1. Egg- Rs.25")
        print("2. Tea- Rs.40")
        print("3. Toast- Rs.50")
        
        b_choice = int(input("Enter your breakfast choice(1-3):"))
        
        match b_choice:
            case 1:
                print("You have ordered egg.")
            case 2:
                print("You have ordered tea.")
            case 3:
                print("You have ordered toast.")
            case _:
                print("Invalid breakfast choice. Please choose again.")
    case 2:
        print("You have ordered lunch.")
    case 3:
            print("You have ordered snacks.")
    case 4:
            print("You have ordered dinner.")
    case 5:
            print("You have ordered desserts.")
    case 6:
            print("You have ordered drinks.")
    case _:
            print("Please enter valid number(1-6). Try again.")
    
            
```

### Break and Continue

```python
# break-used to stop program forcefully
# continue-used to skip specific condition
```

```python
for i in range(1,11):
    if i ==5:
        continue 
    # here we skip 5 and run other parts normally
    print(i)
```

```python
for i in range(1,11):
    if i ==5:
        break
    # here we stop program from running when i=5
    print(i)
```
