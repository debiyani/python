## Dictionary 

### Add, Update, Delete

```python
person_details={
    'Name':'Bob',
    'Age':40,
    'Address':'Kathmandu',
    'Phone':'9876543210'
}
person_details['Job position']='Data Engineer' #this way we can only add key-value pair and not value-only 
person_details
# to add value only we use list_person_details

```

```python
list_person_details={
    'Name':['Bob','Ram','Shyam'],
    'Age':[20,23.30],
    'Address':['Kathmandu','Biratnagar','Lalitpur']
}
list_person_details['Name'].insert(4, "Taylor")
list_person_details
```

### Update

```python
person_details['Name']='John'
person_details
```

#### Updating list dictionary

```python
list_person_details['Name'][2]='John'
list_person_details
```

### Delete

```python
''' 
-> del dict[key]
-> pop(key)
->popitem()  ->removes data from last
'''
```

```python
print(f"Before deletion:{person_details}")
del person_details['Name']
print(person_details)
```

```python
list_person_details.pop('Age')
# list_person_details['Address'].pop()
person_details
```

```python
list_person_details.popitem()
```

```python
# 3 functions for printing a dict
'''
->keys()
->values()
->items()
'''
```

```python
for key in person_details.keys():
    print(key)
```

```python
for val in person_details.values():
    print(val)
```

```python
for key, val in person_details.items():
    print(f"{key}: {val}")
```

#### For printing list dictionary

```python
list_person_details={
    'Name':['Bob','Ram','Shyam'],
    'Age':[20,23.30],
    'Address':['Kathmandu','Biratnagar','Lalitpur']
}
list_person_details['Name'].insert(4, "Taylor")
list_person_details
```

```python
for key, value in list_person_details.items():
    print(f"{key}: {val}")
    for idx, val in enumerate(value):
        print(f"{idx}: {val}")
```

```python
gradebook = {"Sam": [80, 85, 90], 
             "Tara": [92, 88, 95]}
gradebook['Leo']=[70,75,80]
gradebook['Sam'][1]=88
gradebook['Tara'].pop()
for key, val in gradebook.items():
    print(f"{key}: {max(val)}")
```
