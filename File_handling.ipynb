### CSV

```python
import csv
```

```python
# multi-d list (list to csv)
products = [
    ['Name', 'Brand','Price'],
    ['Laptop','HP',150000],
    ['Watch','NavyForce',35000]
]
# must need multi d list
```

### Create File

```python
# open(filename, operation, newLine)
with open('products.csv','w',newline='') as file:
    writer=csv.writer(file)
    writer.writerows(products)
```

```python
with open('products.csv','r') as file:
    reader=csv.reader(file)
    for row in reader:
        print(row)
```

```python
new_products=[['Bag','Bello',5000]]
```

```python
with open('products.csv','a',newline='') as file:
    writer=csv.writer(file)
    writer.writerows(new_products)
```

### Dictionary to CSV 

```python
dict_product=[
    {
        'Name':'Laptop',
        'Brand':'HP',
        'Price':150000
    },
    {
            'Name':'Watch',
            'Brand':'NavyForce',
            'Price':35000
        },
    {
            'Name':'Bag',
            'Brand':'Bello',
            'Price':5000
        }
]
# pass json data
```

```python
with open('dict_product.csv','w',newline='') as file:
    columns=['Name','Brand','Price']
    writer=csv.DictWriter(file, fieldnames=columns)
    writer.writeheader()
    writer.writerows(dict_product)
```

```python
with open('dict_product.csv','r') as file:
    reader=csv.DictReader(file)
    for row in reader:
        for key,val in row.items():
            print(f'{key}: {val}')
        print('\n')
```

```python
# append
new_dict_product=[
    {
        'Name':'Shoe',
        'Brand': 'Caliber',
        'Price':5000,
        'Stock':12
    }
]
```

```python
with open('dict_product.csv','a',newline='') as file:
    columns=['Name','Brand','Price','Stock']
    writer=csv.DictWriter(file, fieldnames=columns)
    writer.writerows(new_dict_product)
```

### JSON( JavaScript Object Notation)

```python
products_json=[
    {
        'Name':'Laptop',
        'Brand':'HP',
        'Price':150000
    },
    {
            'Name':'Watch',
            'Brand':'NavyForce',
            'Price':35000
        },
    {
            'Name':'Bag',
            'Brand':'Bello',
            'Price':5000
        }
]
```

### Create File

```python
import json
```

```python
with open('produucts.json','w') as file:
    json.dump(products_json, file, indent=4)
    # indent just for formatting
```

### Read File

```python
with open('produucts.json','r') as file:
    data = json.load(file)
    for row in data:
        for key, val in row.items():
            print(f'{key}: {val}')
        print('\n')
```
