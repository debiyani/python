### Multiples Skipper
    Write a program that loops through numbers from 1 up to 30. If a number is a multiple of 3, skip 
    printing it entirely using the continue statement. Print all other numbers.

```python
for i in range(1,30):
    if i%3==0:
        continue
    print(i)
```

### Factorial Machine
    Write a loop that calculates the mathematical factorial of a given variable n. (For example, if n = 
    5, your loop should calculate 5 x 4 x 3 x 2 x 1 and print 120).

```python
fact=1
n=int(input("Enter a number:"))
for i in range (1,n+1):
    fact *=i
print(fact)
```

### Number Pattern Grid
    Write nested loops (a loop inside another loop) to print out a right-angle triangle pattern of 
    increasing numbers up to 4 rows exactly like this:
    1
    12
    123
    1234

```python
for i in range(1,5):
    for j in range(1,i+1):
        print(j, end=" ")
    print()
```

### Digit Sum Calculator
    Write a while loop that takes a multi-digit integer variable (e.g., num = 456) and calculates the 
    sum of its individual digits (4 + 5 + 6 = 15). Note: You must use arithmetic operators (% and //). 
    Do not convert the number to a string.

```python
num = int(input("Enter multi-digit integer variable."))
sum=0
while num>0:
    number=num%10
    sum+=number
    num=num//10
# here num%10 helps us to extract rightmost digit so we can sum and num//10 removes rightmost digit
print(f"The sum of all digits is: {sum}")
```

### Prime Number Verifier
    Write a program that takes an integer variable n and uses a loop to check if it is a prime number 
    (divisible only by 1 and itself). Print "Prime" if it is prime, and "Not Prime" if it is not.

```python
is_prime=True
n=int(input("Enter a number:"))
for i in range(2,n):
    if n%i==0:
        is_prime=False
if is_prime:
    print("Prime")
else:
    print("Not Prime")
```
