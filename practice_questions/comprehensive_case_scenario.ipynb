### E-commerce Dynamic Shipping Engine

```python
cart_total = float(input("Enter your cart total:"))
delivery_region = input("Enter your geographical delivery region: Domestic or International")
membership_status=input("Enter your membership status:")
if cart_total>150:
    shipping_cost=0
    print("You have free shipping.")
else :
    if delivery_region=="domestic":
        shipping_cost = 10
    elif delivery_region == "international":
        shipping_cost=25
    else:
        print("Please enter valid geographical destination.")
    if membership_status=="premium":
        discount =5
        shipping_cost= shipping_cost-discount
        print(f"Discount for premium membership: {discount}")
    
print(f"Shipping cost: {shipping_cost}")
print(f"Your total: {shipping_cost+cart_total}")
```
