### Case Study 4: The Digital Savings Goal Tracker 
    Scenario: 
    A financial app helps users save up money for a specific goal (like buying a laptop). The user types in their total target goal amount. The app starts their savings balance at 0 and tracks individual monetary deposits. 
    The business rules of the app state: 
    • The app keeps asking the user to type an amount to deposit as long as their balance is below the target. 
    • If a user tries to enter a negative number (e.g., -50), the app must print "Error: Cannot deposit negative funds!" and ignore that entry. 
    • If a single deposit matches or exceeds the remaining amount needed, a nested check should determine if they hit the target exactly or went over it, printing a customized celebratory message for each scenario before ending the tracking session. 
    Your Task: 
    Write a program that takes a target savings goal amount from the user. Use a loop that runs while the total savings are less than the goal. Use conditionals inside the loop to validate the deposit entries and manage the targeted goal milestones.

```python
total_target= float(input("Enter your total target goal amount:"))
balance =0
remaining_amount=total_target
while (balance<=total_target):
    deposit=float(input("Enter your deposit to reach your target goal:"))
    if deposit<0:
        print("Error: Cannot deposit negative funds!")
        continue
    if deposit>=remaining_amount:
        ans=input("Did you enter the right amount?: yes or no")
        if ans.lower()=="yes":
            print("Great")
        else:
            deposit=float(input("Enter your deposit to reach your target goal:"))
    balance+= deposit
    remaining_amount=total_target-balance
print("You reached you target goal")
```
