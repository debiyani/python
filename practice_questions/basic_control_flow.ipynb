### Simple Gatekeeper
    Write a program that takes an integer variable named age. If the age is 18 or older, print"Allowed to vote". Otherwise, print "Too young to vote".


```python
age = int(input("Enter age:"))
if age>=18:
    print("Allowed to vote.")
else:
    print("Too young to vote.")
```

### Positive, Negative, or Zero
    Write a script that checks a number variable named num. Print "Positive" if it is greater than zero, "Negative" if it is less than zero, and "Zero" if it equals zero exactly.

```python
num = int(input("Enter a number:"))
if num > 0:
    print("Positive")
elif num < 0:
    print("Negative")
else:
    print("Zero")
```

### Movie Ticket Pricing
    Write a program that determines movie ticket categories based on an age variable:
    • Under 5 years old: print "Free"
    • 5 to 17 years old: print "10"
    • 18 years and older: print "15

```python
age =int(input("Enter your age:"))
if age<5:
    print("Free")
elif age>=5 and age<=17:
    print("10")
else:
    print("15")
```

### Triangle Type Checker
    Given three separate integer side lengths s1, s2, and s3, write a program using an if-else
    condition to print "Equilateral" if all three sides are perfectly equal, or "Not Equilateral" if they are not.

```python
s1 = int(input("Enter side length s1:"))
s2 = int(input("Enter side length s2:"))
s3 = int(input("Enter side length s3:"))
if s1==s2 and s1==s3:
    print("Equilateral")
else:
    print("Not Equilateral")
```

### Nested Login Check
    Write a program that uses nested if statements to verify a login. First, check if a variable
    username matches "admin". If it does, check if a variable password matches "secure123".
    • If both match, print "Access Granted".
    • If the username is right but the password is wrong, print "Wrong Password".
    • If the username is wrong, print "User Not Found"

```python
username=input("Enter username:")
password=input("Enter password:")
if username=="admin":
    if password=="secure123":
        print("Access granted")
    else:
        print("Wrong password")
else:
    print("User Not Found")
```
