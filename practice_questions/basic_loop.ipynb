### Rocket Countdown
    Write a while loop that counts down from 5 to 1, prints each number, and then prints the 
    message "Blast off!" once the loop finishes.

```python
count = 5
while count>=1:
    print(count)
    count-=1
print("Blast off!")
```

### Factor Counter
    Write a program that takes an integer variable N and uses a loop to find and print all of its 
    mathematical factors (any number that divides N perfectly with no remainder).

```python
N = int(input("Enter a number:"))
for i in range(1,N):
    if N%i==0:
        print(i)
```

### Running Total Accumulator
    Write a program using a while True infinite loop that continually prompts the user to input a 
    positive integer to add to a total_sum variable. The loop must terminate immediately using a 
    control statement if the user inputs any negative number.

```python
total_sum=0
while True:
    positive_integer=int(input("Enter a positive integer:"))
    total_sum+=positive_integer
    print(total_sum)
    if positive_integer<0:
        break
    # break is a control statement 
```

### String Character Spacer
    Write a for loop that takes a string variable (e.g., text = "PYTHON") and iterates through it to 
    print out each character followed immediately by a hyphen on the same line (e.g., P-Y-T-H-O-N-).

```python
text=input("Enter a string:")
for i in text:
    print(i, end ="-")
```
