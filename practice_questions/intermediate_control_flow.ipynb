### The Discount Calculator
    A shop offers discounts based on a customer's total purchase amount and their boolean 
    membership status (is_member):
    • If the purchase is over 100 and they are a member, apply a 20% discount.
    • If the purchase is over 100 but they are not a member, apply a 10% discount.
    • If the purchase is 100 or less, apply no discount regardless of membership.
    Write a program using nested if statements to calculate and print the final price.

```python
total_purchase=float(input("Enter your total purchase:"))
is_member=input("Enter your membership status: yes or no")
if total_purchase>100:
    if is_member=="yes":
        discount = 0.20*total_purchase
        print("20% discount on purchase over 100 for members.")
    else:
        discount=0.10*total_purchase
        print("10% discount on purchase over 100 for non-members.")
if total_purchase<=100:
    discount=0
    print("No discount for purchase less than or equal to 100.")
final_price=total_purchase-discount
print(f"Your total purchase: {total_purchase}")
print(f"Discount: {discount}")
print(f"Final price: {final_price}")
```
