### Case Study 3: The Industrial Furnace Temp Controller 
    Scenario: 
    A manufacturing plant runs a chemical heating furnace. The furnace starts at an initial temperature of 200°C. The operator needs to safely heat the furnace up to a target temperature of 300°C. 
    The control logic loop behaves as follows: 
    1. The program uses a loop to heat the furnace, increasing the temperature by 15°C at every step. 
    2. At each step, a pressure check is evaluated via user input. If the operator types that the pressure is "HIGH", the furnace cannot safely absorb more heat; the program must skip the temperature increase for that specific step and check again. 
    3. If the temperature ever spikes and exceeds 320°C, an automatic safety vent must trigger, forcing the loop to stop entirely to prevent an explosion. 
    Your Task: 
    Write a program using a loop that runs as long as the temperature is less than 300. Inside the loop, ask the operator for the pressure status ("NORMAL" or "HIGH"). Use conditions to handle skipping the temperature adjustment or breaking out if a safety limit is breached.
