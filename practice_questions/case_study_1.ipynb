#### Case Study 1: The Smart Crop Irrigation System 
    Scenario: 
    An automated greenhouse uses a smart valve system to water crops based on real-time sensor 
    readings. The system runs a 7-day testing cycle. 
    The watering logic rules are: 
    • If the moisture is below 30%, turn the water valve "ON" at full capacity. 
    • If the moisture is between 30% and 60% (inclusive), turn the water valve "ON" at half 
    capacity. 
    • If the moisture is above 60%, keep the water valve "OFF". 
    Safety Override: If the moisture reading ever hits exactly 100%, the system assumes the sensor 
    is submerged or broken. It must immediately print a "Sensor Error: Emergency Shutdown" 
    alert and stop the entire weekly loop right away. 
    Your Task: 
    Write a program that uses a loop to count through days 1 to 7. Inside the loop, prompt the user to 
    input the moisture reading for that day. Use conditionals to determine the valve status and apply 
    the emergency override if the error condition is met.

```python
for i in range (1,8):
    moisture_reading=int(input(f"Enter moisture reading for day{i}:"))
    if moisture_reading<30:
        print("ON at full capacity.")
    elif moisture_reading>=30 and moisture_reading<=60:
        print("ON at half capacity.")
    elif moisture_reading>60 and moisture_reading<100:
        print("OFF")
    else:
        print("Sensor Error: Emergency Shutdown")
        break
```
