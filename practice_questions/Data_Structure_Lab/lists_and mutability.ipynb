#### Lab Question 1:
    The Personal Inventory Manager Create a list named inventory containing the following items: "Laptop", "Mouse", and "Keyboard". Perform the following operations: 
    1. Add "Monitor" to the end of the list using append(). 
    2. Insert "Webcam" at the second position (index 1) in the list. 
    3. Sort the list alphabetically. 
    4. Remove "Mouse" from the list using its name. 
    5. Pop the last item from the list and print the final state of the inventory. 

```python
inventory=["Laptop", "Mouse","Keyboard"]
inventory.append("Monitor")
inventory.insert(1,"Webcam")
inventory.sort()
inventory.remove("Mouse")
inventory.pop()
print(inventory)
```

#### Lab Question 2:
    Advanced Slicing & Analysis Given the list: numbers = [12, 45, 2, 89, 33, 10, 55, 78, 90, 6] 
    1. Print a slice of the list containing only the middle four elements. 
    2. Use slicing to print the list in reverse order. 
    3. Extract every third element starting from the first one. 
    4. Find and print the length, maximum value, and minimum value of this list using 
    Python’s built-in functions.

```python
numbers=[12,45,2,89,33,10,55,78,90,6]
print(numbers[3:7])
print(numbers[::-1])
print(numbers[::3])
print(f"Length: {len(numbers)}")
print(f"Maximum: {max(numbers)}")
print(f"Minimum: {min(numbers)}")
```
