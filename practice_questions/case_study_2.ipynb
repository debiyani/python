### Case Study 2: The Ticket Booth & Group Discounter 
    Scenario: 
    An amusement park ticket booth calculates prices for incoming groups. The booth agent processes a group by entering the age of each individual family member one by one. 
    The base rules for ticket pricing are: 
    • Age 12 and under: 10 
    • Age 13 to 64: 20 
    • Age 65 and older: 15 
    Group Dynamics: If the agent encounters a child under the age of 3, that individual ticket is  free, but the program must also trigger a special rule: it must remind the agent to print a "Free Stroller Pass" voucher for that turn. 
    Your Task: 
    Write a program that asks the user how many people are in their group. Use a loop to iterate exactly that many times. During each iteration, ask for the age of the person, calculate the individual ticket cost, handle the nested stroller voucher rule, and keep a running total of the entire group's cost.

```python
group = int(input("How many people in the group?:"))
group_cost=0
for i in range(1,group+1):
    age=int(input("What is the age of the person?:"))
    if age<3:
        print("Free Stroller Pass")
        print("Ticket price for age under 3: 0")
        ticket_price=0
    elif age<=12:
        print("Ticket price for age 12 and under: 10")
        ticket_price=10
    elif age<=64:
        print("Ticket price for age 13 to 64: 20")
        ticket_price=20
    else:
        print("Ticket price for 65 and older: 15")
        ticket_price=15
    group_cost+=ticket_price
print(f"The total cost for the group is: {group_cost}")
    
    
```
