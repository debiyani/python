## Lambda Functions

```python
# Normal function
def calculate(a,b):
    add =a+b
    print(add)
calculate(5,4)
```

```python
# Lambda function Syntax:
# var_name = lambda arg1, arg2 : return 'value'
# lambda keyword converts the var_name into a function
```

```python
addition = lambda a,b : print(a+b)
```

```python
addition(6,3)
```

```python
odd_even = lambda num :f"{num}: Even" if num%2==0 else f"{num}: Odd"
# can only use if else and not elif in lamda function
# can use for loop 
```

```python
odd_even(3)
```

```python
data=lambda : [f"{i}: Even" if i%2==0 else f"{i}: Odd" for i in range(1,21)]
# to use for loop we must wrap it with []list, or tuple()
```

```python
data()
```

```python
prices=[100,250,75,400,120,310]
cheap_items=[]
expensive_items=[]
expensive_cheap= lambda:[expensive_items.append(price) if price>=200 else cheap_items.append(price) for price in prices]
```

```python
expensive_cheap()
print(expensive_items)
print(cheap_items)
```

```python
expensive=lambda : [price for price in prices if price>=200] 
#here since we dont have else condition we use var price in the start
cheap=lambda : [price for price in prices if price<200]
# lambda functions make program faster
print(expensive())
print(cheap())
```
