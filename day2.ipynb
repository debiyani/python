```python
# student pass or fail
marks=int(input('Enter your marks:'))

if(marks >= 40):
    print('Pass')
else:
    print('Fail')
```

```python
# temperature check (hot,warm or cold)
temperature=float(input("Enter temperature:"))
if temperature>=28:
    print("It's hot today.")
elif temperature>=15 and temperature<28:
    print("It's warm today.")
else:
    print("It\'s cold today.") 
# \' escape sequence for '
```

#### Income tax calculator

```python
income = float(input('Enter annual income:'))
if income>=0 and income<=25000:
    tax_amount=0
    remaining_balance=income-tax_amount
elif income>25000 and income<=50000:
    tax_amount= income*(5/100) 
    # can also do tax_amount = income*0.05
    remaining_balance=income-tax_amount

elif income>50000 and income<=100000:
    tax_amount= income*(20/100)
    remaining_balance=income-tax_amount
elif income >100000:
    tax_amount= income*(30/100)
    remaining_balance=income-tax_amount
    
else:
    print("Invaid income. Please enter valid income.")
print(f"Total income:{income}")
print(f"Tax amount:{tax_amount}")
print(f"Remaining balance:{remaining_balance}")





```

#### Login authenticator

```python
default_username= "root"
default_password = "password"
username = input("Enter username:")
password =input("Enter password:")
if username == default_username and password == default_password:
    print("Login successfull.")
elif username == default_username and password != default_password:
    print("Incorrect password.")
elif username != default_username and password== default_password:
    print("User not found.")
else:
    print("Username or password incorrect.")
```

#### ATM Withdrawl System

```python
# Nested loop
pin = "1234"
balance = 10000
entered_pin = input("Enter your pin:")
withdraw_balance = int(input("Enter balance to withdraw:"))
if (pin==entered_pin):
    if (withdraw_balance<=balance):
        if (withdraw_balance%1000==0):
            new_balance=balance-withdraw_balance
            print(f"Balance before withdrawl:{balance}")
            print(f"Rs.{withdraw_balance} successfully withdrawn.")
        else:
            print("Amount must be multiple of 1000.")
    else:
        print("Balance not sufficient")
else:
    print("Incorrect pin.")
print(f"Current balance: Rs.{new_balance}")
```
