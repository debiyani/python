### Default Arguments

```python
def employeeDetails(emp_id, emp_name, emp_position='Unknown position', emp_job_type='Unknown type'):
    print(f"Employee ID: {emp_id}")
    print(f"Employee Name: {emp_name}")
    print(f"Employee Position: {emp_position}")
    print(f"Employee Job Type: {emp_job_type}")

employeeDetails(1,'Ram', 'Intern')
# the unknown type ie default value must be set at last like emp_id="unknown", emp_name wont work as emp name doesnt have default value
```

### Keyword arguments

```python
def employeeDetails(emp_id="unknown", emp_name="unknown", emp_position="unknown", emp_job_type="unknown"):
    print(f"Employee ID: {emp_id}")
    print(f"Employee Name: {emp_name}")
    print(f"Employee Position: {emp_position}")
    print(f"Employee Job Type: {emp_job_type}")
# keyword arguments used to remove errors caused by default arguments
#  allow you to pass values into a Python function by explicitly specifying the parameter name in the format name=value
```

```python
employeeDetails(2, 'Ram','Full-time')
```

# Args and Kwargs(All arguments and All keywords)

```python
# all arguments -> *args (type tuple used for data)
# all keyword arguments -> **kwargs (has key value pair type dict)
# kwargs used in forms
# args used in ai/ml, data science and data analysis
```

```python
# *args
def empDetails(*employees):
    print(employees)
    print(type(employees))

empDetails('Ram', 'Kathmandu','Masters',34)
```

```python
def empDetailskwargs(**kwargs):
    print(kwargs)
    for key, val in kwargs.items():
        print(f"{key}: {val}")
    print(type(kwargs))
```

```python
empDetailskwargs(Name='Ram', Address='Kathmandu', Qualification='Masters', Age='34', abc=33)
```
