```python
from selenium import webdriver
from selenium.webdriver.common.by import By
```

```python
# driver = webdriver.Edge() 
# for using edge
```

```python
driver = webdriver.Chrome()
```

```python
website="https://merolagani.com/LatestMarket.aspx"
```

```python
driver.get(website)
```

```python
table_body = driver.find_element(By.TAG_NAME, 'tbody')
```

```python
table_row = table_body.find_elements(By.TAG_NAME, 'tr')
```

```python
symbol_col=[]
ltp_col=[]
change_col=[]
open_col=[]
high_col=[]
low_col=[]
qty_col=[]
pclose_col=[]
diff_col=[]
```

```python
# for row in table_row:
#     columns=row.find_elements(BY.TAG_NAME,'td')
#     for col in columns:
#         print(col.text)
```

```python
for row in table_row:
    columns=row.find_elements(By.TAG_NAME,'td')
        
    symbol_col.append(columns[0].text)
    ltp_col.append(columns[1].text)
    change_col.append(columns[2].text)
    open_col.append(columns[3].text)
    high_col.append(columns[4].text)
    low_col.append(columns[5].text)
    qty_col.append(columns[6].text)
    pclose_col.append(columns[7].text)
    diff_col.append(columns[8].text)
```

```python
symbol_col
```

```python
data={
    "Symbol":symbol_col,
    "LTP":ltp_col,
    "%change":change_col,
    "Open":open_col,
    "High":high_col,
    "Low":low_col,
    "QTY":qty_col,
    "PClose":pclose_col,
    "Diff":diff_col
}
```

```python
driver.quit()
```

```python
import pandas as pd
```

```python
df=pd.DataFrame(data)
```
