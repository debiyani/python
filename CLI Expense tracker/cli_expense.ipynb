```python
expense=0
monthly_budget = input(int("Enter your monthly budget."))
```

```python
class ExpenseMoreThanBudgetError:
    pass
class BudgetNotValidError:
    pass
```

```python
class expense_tracker:
    def __init__(self,expense_name, expense_amount):
        self.expense_name=expense_name
        self.expense_amount=expense_amount
        
    def add_expense(self):
        try:
            if self.expense_amount<monthly_budget:  
                expense += self.expense_amount
        except ExpenseMoreThanBudgetError as ebe:
            print(ebe)
        
    def edit_expense(self):
        self.expense_amount = int(input("Enter edited expense."))
        
    def delete_expense(self):
        expense-= self.expense_amount
```
